tengo un pdf en esta url, haz un programa en python para extraer los datos

```python
import requests
import fitz  # PyMuPDF

def descargar_pdf(url, nombre_archivo):
    response = requests.get(url)
    response.raise_for_status()  # Para asegurarnos que la descarga fue exitosa
    with open(nombre_archivo, 'wb') as f:
        f.write(response.content)
    print(f"PDF descargado como {nombre_archivo}")

def extraer_texto_pdf(nombre_archivo):
    texto_completo = ""
    documento = fitz.open(nombre_archivo)
    for pagina in documento:
        texto_completo += pagina.get_text()
    return texto_completo

if __name__ == "__main__":
    url_pdf = "https://www.upt.edu.pe/upt/sgc/assets/ckeditor/kcfinder/upload/files/RESULTADOS_ORDINARIO_2025-I.pdf"  # Cambia esta URL por la del PDF que quieres
    archivo_pdf = "archivo_descargado.pdf"

    descargar_pdf(url_pdf, archivo_pdf)
    texto = extraer_texto_pdf(archivo_pdf)
    print("Texto extraído del PDF:")
    print(texto)
```

hazme un programa en python para buscar con el nombre de una persona todo lo que se pueda extraer de internet, usa google dorks, extrae fotos, perfiles, presentaciones y redes sociales entre otros

```python
import webbrowser
from googlesearch import search
import time

def generar_dorks(nombre):
    nombre_comillas = f'"{nombre}"'
    return {
        "Perfiles de LinkedIn": f'site:linkedin.com/in {nombre_comillas}',
        "Perfiles de Facebook": f'site:facebook.com {nombre_comillas}',
        "Imágenes": f'{nombre_comillas} site:images.google.com',
        "Presentaciones PDF": f'filetype:pdf {nombre_comillas}',
        "Presentaciones PPT": f'filetype:ppt {nombre_comillas}',
        "GitHub o GitLab": f'{nombre_comillas} site:github.com OR site:gitlab.com',
        "Twitter / X": f'{nombre_comillas} site:x.com OR site:twitter.com',
        "Instagram": f'{nombre_comillas} site:instagram.com',
    }

def buscar_google(query, num_results=5):
    try:
        resultados = list(search(query, num_results=num_results))
        return resultados
    except Exception as e:
        print(f"Error al buscar '{query}': {e}")
        return []

def ejecutar_busqueda(nombre):
    dorks = generar_dorks(nombre)
    for descripcion, query in dorks.items():
        print(f"\n🔍 {descripcion} — Buscando: {query}")
        resultados = buscar_google(query)
        for r in resultados:
            print(f"  👉 {r}")
        time.sleep(2)  # Para evitar bloqueo por rate-limit

def abrir_en_navegador(nombre):
    dorks = generar_dorks(nombre)
    for descripcion, query in dorks.items():
        url = f"https://www.google.com/search?q={query.replace(' ', '+')}"
        print(f"🧭 Abriendo en navegador: {descripcion}")
        webbrowser.open_new_tab(url)
        time.sleep(1)

if __name__ == "__main__":
    nombre_persona = input("🔎 Ingresa el nombre de la persona: ")
    print("\n--- RESULTADOS EN CONSOLA ---")
    ejecutar_busqueda(nombre_persona)
    print("\n--- ABRIENDO BÚSQUEDAS EN EL NAVEGADOR ---")
    abrir_en_navegador(nombre_persona)
```
